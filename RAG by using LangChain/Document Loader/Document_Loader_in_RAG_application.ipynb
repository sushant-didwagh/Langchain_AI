#Text Loader

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader('cricket.txt', encoding='utf-8')
docs = loader.load()
print(type(docs))

print(docs)
print(len(docs))
#print(docs[0])
print(type(docs[0]))
print(docs[0].page_content)
print(docs[0].metadata)
```

```python
from re import template

from groq import Groq
client = Groq(api_key="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")
from langchain_groq import ChatGroq
llm = ChatGroq(groq_api_key=client.api_key, model_name="llama-3.1-8b-instant")

from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    template='write summary for the following poem \n {poem}',
    input_variables=['poem']
)
parser = StrOutputParser()

chain= prompt|llm|parser
chain.invoke({'poem':docs[0].page_content})
```

#DirectoryLoader

```python
from langchain_community.document_loaders import DirectoryLoader
loader = DirectoryLoader(
    path='books',
    glob='*.pdf',
    loader_cls=PyPDFLoader,
)
docs3 = loader.load()
print(docs3)

docs4 = loader.lazy_load() #best, fast loading
```
