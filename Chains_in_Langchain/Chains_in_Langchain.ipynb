#Conditional Chain


```python

from groq import Groq
client = Groq(api_key="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")

from langchain_groq import ChatGroq
llm = ChatGroq(groq_api_key=client.api_key, model_name="llama-3.1-8b-instant")


from langchain_core.output_parsers import StrOutputParser, PydanticOutputParser
parser1 = StrOutputParser()

from pydantic import BaseModel, Field # for structured Consistent output
from typing import Literal
class feedback(BaseModel):
  sentiment: Literal['positive', 'negative'] = Field(description='Give the sentiment of the feedback')
parser2 = PydanticOutputParser(pydantic_object= feedback)

from langchain_core.prompts import PromptTemplate
prompt1=PromptTemplate(
    template='Classify the sentiment of the following text into positive or negative.\n\nTEXT: {feedback}\n\n{format_instruction}\n\nRespond ONLY with the JSON object. The sentiment value should be a string, not a list. Do NOT include any other text, explanations, or code blocks.',
    input_variables=['feedback'],
    partial_variables={'format_instruction': parser2.get_format_instructions()} #output structure is alsways same
)


classifier_chain = prompt1 | llm | parser2


prompt2=PromptTemplate(
    template='write appropriate answer to this positive feedback \n {feedback}',
    input_variables=['feedback']
)
prompt3=PromptTemplate(
    template='weite appropriate answer to this negative feedback \n {feedback}',
    input_variables=['feedback']
)

from langchain_core.runnables import RunnableParallel, RunnableBranch, RunnableLambda
branch_chain = RunnableBranch(
    (lambda x: x.sentiment == 'positive', prompt2|llm|parser),
    (lambda x: x.sentiment == 'negative', prompt3|llm|parser),
    RunnableLambda(lambda x: "could not find sentiment") #lambda function to runnable chain so we can use it as a another chain
)

f_chain = classifier_chain | branch_chain
result = f_chain.invoke({'feedback': 'this is terrible smartphone'})
print(result)
```
