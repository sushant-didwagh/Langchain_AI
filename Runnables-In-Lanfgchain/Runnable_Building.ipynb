#Chains Before Runnable

```python
import random
class NakliLLM:
  def __init__(self):
    print('LLM created')

  def predict(self, prompt):
    response_list=[
        'Delhi is capital of indial',
        'virat is best cricketer',
        'PCCOE is one of the demanded collage of Pune'
    ]

    return {'response': random.choice(response_list)}
```

```python
llm = NakliLLM()
```

```python
llm.predict('what is capital of india')
```

```python
class NakliPromptTemplate:
  def __init__(self, template, input_variables):
    self.template = template
    self.input_variables = input_variables

  def format(self, input_dict):
    return self.template.format(**input_dict)
```

```python
template=NakliPromptTemplate(
    template='write a {length} poes about {topic}',
    input_variables=['length','topic']
)
```

```python
prompt = template.format({'length': 'short','topic': 'india'})
prompt
```

```python
llm.predict(prompt)
```

```python
class NakliLLMChain():

  def __init__(self, llm, prompt):
    self.llm = llm
    self.prompt = prompt

  def run(self, input_dict):
    final_prompt = self.prompt.format(input_dict)
    result = self.llm.predict(final_prompt)

    return result['response']
```

```python
template=NakliPromptTemplate(
    template='write a {length} poes about {topic}',
    input_variables=['length','topic']
)
```

```python
llm = NakliLLM()
```

```python
chain = NakliLLMChain(llm, template)
```

```python
chain.run({'length':'short', 'topic':'modi'})
```

#Runnables Creation


```python
from abc import ABC, abstractmethod
```

```python
class Runnables(ABC):
  @abstractmethod
  def invoke(input_data):
    pass
```

```python
import random
class NakliLLM(Runnables):
  def __init__(self):
    print('LLM created')

  def invoke(self, prompt):
    response_list=[
        'Delhi is capital of indial',
        'virat is best cricketer',
        'PCCOE is one of the demanded collage of Pune'
    ]

    return {'response': random.choice(response_list)}

  def predict(self, prompt):
    response_list=[
        'Delhi is capital of indial',
        'virat is best cricketer',
        'PCCOE is one of the demanded collage of Pune'
    ]

    return {'response': 'call invoke insted of predict'}
```

```python
llm=NakliLLM()
```

```python
class NakliPromptTemplate(Runnables):
  def __init__(self, template, input_variables):
    self.template = template
    self.input_variables = input_variables

  def invoke(self, input_dict):
    return self.template.format(input_dict)

  def format(self, input_dict):
    return self.template.format('call invoke insted of format')
```

```python
class RunnableStrOutputParser(Runnables):
  def __init__(self):
    pass

  def invoke(self, input_data):
    return input_data['response']
```

```python
class RunnableConnector(Runnables):
  def __init__(self, runnable_list):
    self.runnable_list = runnable_list

  def invoke(self, input_data):
    for runnable in self.runnable_list:
      input_data = runnable.invoke(input_data)

    return input_data
```

```python
parser = RunnableStrOutputParser()
```

```python
template=NakliPromptTemplate(
    template='write a {length} poes about {topic}',
    input_variables=['length','topic']
)
```

```python
chain = RunnableConnector([template, llm, parser])

chain.invoke({'length': "long", 'topic': 'india'})
```

```python
template1=NakliPromptTemplate(
    template='write a joke about {topic}',
    input_variables=['topic']
)
```

```python
template2=NakliPromptTemplate(
    template='explain the following joke{response}',
    input_variables=['response']
)
```

```python
llm=NakliLLM()
```

```python
parser = RunnableStrOutputParser()
```

```python
chain1 = RunnableConnector([template1, llm])
chain2 = RunnableConnector([template2, llm, parser])

final_chain = RunnableConnector([chain1, chain2])
final_chain.invoke({'topic': 'Sex'})
```
