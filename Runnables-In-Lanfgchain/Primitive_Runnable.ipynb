#RunnableParallel

```python
from langchain_core.runnables import  RunnableParallel
prompt1=PromptTemplate(
    template='generaet twit about {topic}',
    imput_variables=['topic']
)

prompt2=PromptTemplate(
    template='generaet linkedin post about {topic}',
    imput_variables=['topic']
)

parser=StrOutputParser()

p_chain= RunnableParallel({
    'tweet': RunnableSequence(prompt1,model1,parser),
    'linkedin': RunnableSequence(prompt2,model1,parser)
})
p_chain.invoke({'topic':'AI'})
```

#RunnabelPassThrough
input is as it is output


```python
from langchain_core.runnables import RunnablePassthrough
joke_gen_chain = RunnableSequence(prompt1,model1,parser)

parallel_chain = RunnableParallel({
    'joke': RunnablePassthrough(),
    'explanation': RunnableSequence(prompt2,model1,parser)
})

final_chain = RunnableSequence(joke_gen_chain,parallel_chain)
final_chain.invoke({'topic':'AI'})
```

#RunnabelLambda

```python
from langchain_core.runnables import RunnableLambda, RunnableSequence

def word_count(text):
  return len(text.split())

joke_gen = RunnableSequence(prompt1,model1,parser)
par_chain = RunnableParallel({
    'joke': RunnablePassthrough(),
    'word_count': RunnableLambda(word_count) # can be 'word_count': RunnableLambda(lambda x: len(x.split()))
})

final_chain2 = RunnableSequence(joke_gen,par_chain)
final_chain2.invoke({'topic':'AI'})
```
